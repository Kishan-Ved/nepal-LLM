```python
import pandas as pd
import torch
from transformers import AutoTokenizer

# Load the data from the CSV
df = pd.read_csv("/Users/dj/Downloads/all_final_datasets/final_sample_100000.csv")  # Ensure the path and file name are correct
if 'text' not in df.columns:
    raise ValueError("The CSV file must have a 'text' column.")

# Merge all text data from the 'text' column into a single string
merged_articles = df["text"].to_list()
all_merged_articles = " ".join(merged_articles)  # Combine the first 200 rows for processing

# Load the tokenizer
tokenizer_path = "/Users/dj/Downloads/tokenizers/final_sample_150000.csv_SentencePieceBPETokenizer_49152_transformer"
tokenizer = AutoTokenizer.from_pretrained(tokenizer_path)

# Tokenize the combined text
token_ids = tokenizer.encode(all_merged_articles)

# Define context length and pad the token_ids to match the context length
context_length = 512
pad_length = (context_length - len(token_ids) % context_length) % context_length
token_ids.extend([tokenizer.pad_token_id] * pad_length)  # Use tokenizer's `pad_token_id` instead of hardcoded 0

# Reshape the tokenized IDs into a 2D tensor
token_ids_pt = torch.tensor(token_ids, dtype=torch.long).reshape(-1, context_length)

# Create a new DataFrame for inspection if needed
new_df = pd.DataFrame({'input_ids': token_ids_pt.tolist()})

print("Tokenization and saving completed successfully.")
```
